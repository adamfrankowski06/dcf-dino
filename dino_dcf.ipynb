```python
import pandas as pd

rzis = pd.DataFrame({
    "2024": [29_273_792, -22_463_233, 24_185, -4_678_717, -233_653, -14_044, 1_908_330, 6_657, -121_424, 1_793_563, -287_834, 1_505_729],
    "2025": [33_634_155, -25_723_078, 29_231, -5_576_262, -302_944, -19_820, 2_041_282, 8_316, -122_945, 1_926_653, -367_856, 1_558_797],
}, index=[
    "Przychody ze sprzedaży",
    "Koszt własny sprzedaży",
    "Pozostałe przychody operacyjne",
    "Koszty sprzedaży i marketingu",
    "Koszty ogólnego zarządu",
    "Pozostałe koszty operacyjne",
    "Wynik operacyjny",
    "Przychody finansowe",
    "Koszty finansowe",
    "Wynik brutto",
    "Podatek dochodowy",
    "Wynik netto",
])
```

```python
skladniki = ["Przychody ze sprzedaży", "Koszt własny sprzedaży", "Pozostałe przychody operacyjne", "Koszty sprzedaży i marketingu", "Koszty ogólnego zarządu", "Pozostałe koszty operacyjne"]
wynik_operacyjny = rzis.loc[skladniki, "2025"].sum()         #policzyłem wynik operacyjny dla 2025 roku
```

```python
cash_flow = pd.DataFrame({
    "2024": [409_179, -424_123, -55_966, 1_126_352, -1_561_914],
    "2025": [505_031, -466_656, 14_973, 948_166, -2_087_043],
}, index=[
    "Amortyzacja",
    "Zmiana zapasów",
    "Zmiana należności",
    "Zmiana zobowiązań (bez kredytów)",
    "Capex (zakup rzeczowych aktywów trwałych i wartości niematerialnych)",
])

print(cash_flow) #policzyłem cash flow dla 2024 i 2025 roku
```

```python
amortyzacja = cash_flow.loc["Amortyzacja", "2025"]
ebitda = wynik_operacyjny + amortyzacja
print(ebitda) # policzyłem EBITDA dla 2025 roku
```

```python
podatek_dochodowy, wynik_brutto = rzis.loc[["Podatek dochodowy", "Wynik brutto"], "2025"]
esp = podatek_dochodowy / wynik_brutto
if esp < 0:
    esp*= -1
print(esp)  #policzyłem efektywna stope podatkowa dla 2025 roku 
```

```python
nopat = wynik_operacyjny * (1 - esp)
print(nopat) # policzyłem NOPAT dla 2025 roku
```

```python
zmiana_kapitalu_obrtowego = cash_flow.loc[["Zmiana zapasów","Zmiana należności", "Zmiana zobowiązań (bez kredytów)"], "2025"].sum()
fcf = nopat + amortyzacja + zmiana_kapitalu_obrtowego + cash_flow.loc["Capex (zakup rzeczowych aktywów trwałych i wartości niematerialnych)", "2025"]
print(fcf) #policzyłem realna gotówke jaką firma wygenerowala
```

```python
przychod_24,przychod_25 = rzis.loc["Przychody ze sprzedaży", ["2024", "2025"]]
procentowy_wzrost= (przychod_25-przychod_24) / przychod_24
print(procentowy_wzrost) # policzyłem procentowy wzrost przychodów ze sprzedaży w 2025 roku w stosunku do 2024 roku
```

```python
import numpy as np

lata_prognozy = [2026, 2027, 2028, 2029, 2030]
wzrost_start = 0.13
wzrost_koncowy = 0.02

stopy_wzrostu = np.linspace(wzrost_start, wzrost_koncowy, len(lata_prognozy))
print(stopy_wzrostu) # policzyłem stopy wzrostu przychodów ze sprzedaży w latach 2026-2030
```

```python
amortyzacja_jako_procent_przychdow = amortyzacja / przychod_25
print(amortyzacja_jako_procent_przychdow) # policzyłem amortyzacje jako procent przychodów ze sprzedaży w 2025 roku
```

```python
capex_jako_procent_przychdow = cash_flow.loc["Capex (zakup rzeczowych aktywów trwałych i wartości niematerialnych)", "2025"] / przychod_25 * -1
print(capex_jako_procent_przychdow) # policzyłem Capex jako procent przychodów ze sprzedaży w 2025 roku
```

```python
capex_prognoza = np.linspace(capex_jako_procent_przychdow, amortyzacja_jako_procent_przychdow, 5)
print(capex_prognoza)
print(amortyzacja_jako_procent_przychdow) # policzyłem amortyzacje jako procent przychodów ze sprzedaży w 2025 roku
```

```python
przychody_prognoza = []
ostatni_przychod = przychod_25

for r in stopy_wzrostu:
    ostatni_przychod = ostatni_przychod * (1 + r)
    przychody_prognoza.append(ostatni_przychod)

print(przychody_prognoza)# policzyłem prognozowane przychody ze sprzedaży w latach 2026-2030
```

```python
marzz_operacyjna = wynik_operacyjny / przychod_25
print(marzz_operacyjna) # policzyłem marżę operacyjną w 2025 roku
```

```python
przyrost_przychodow = przychod_25 - przychod_24
zmiana_kapitalu_obrtowego_jako_procent_przyrostu_przychdow = zmiana_kapitalu_obrtowego / przyrost_przychodow
print(zmiana_kapitalu_obrtowego_jako_procent_przyrostu_przychdow) # policzyłem zmianę kapitału obrotowego jako procent przyrostu przychodów ze sprzedaży w 2025 roku w stosunku do 2024 roku
```

```python
fcf_5lat = []
przychod_poprzedniego_roku = przychod_25
for i, przychod in enumerate(przychody_prognoza):
    wynik_operacyjny = przychod * marzz_operacyjna
    nopat = wynik_operacyjny * (1 - esp)
    amortyzacja = przychod * amortyzacja_jako_procent_przychdow
    capex = przychod * capex_prognoza[i]
    zmiana_kapitalu_obrtowego = (przychod - przychod_poprzedniego_roku) * zmiana_kapitalu_obrtowego_jako_procent_przyrostu_przychdow
    fcf = nopat + amortyzacja + zmiana_kapitalu_obrtowego - capex
    fcf_5lat.append(fcf)
    przychod_poprzedniego_roku = przychod
print(fcf_5lat) # policzyłem prognozowane FCF w latach 2026-2030
```

```python
rentownosc_pol_obl = 6.3/100
premia_za_ryzyko = 5.33/100
beta_dino = 0.5
koszt_kapitalu_wlasnego = rentownosc_pol_obl + beta_dino * premia_za_ryzyko
print(koszt_kapitalu_wlasnego) # policzyłem koszt kapitału własnego dla Dino
```

```python
kapital_wlasny = 34784592
dlug = 583450
koszt_dlugu = 0.075
wacc = (kapital_wlasny / (kapital_wlasny + dlug)) * koszt_kapitalu_wlasnego + (dlug / (kapital_wlasny + dlug)) * koszt_dlugu * (1 - esp)
print(wacc) #policzyłem wacc
```

```python
zdyskontowane_fcf_5lat = []
for i, fcf in enumerate(fcf_5lat):
    zdyskontowany_fcf = fcf / (1 + wacc) ** (i + 1)
    zdyskontowane_fcf_5lat.append(zdyskontowany_fcf)
print(zdyskontowane_fcf_5lat) #zdyskontowałem prognozowaną gotówke jaką firma wygeneruje w latach 2026-2030
```

```python
ostatni_fcf = fcf_5lat[-1]
tempo_wzrostu = 0.025
stopa_dyskontowa = wacc

tv = ostatni_fcf * (1 + tempo_wzrostu) / (stopa_dyskontowa - tempo_wzrostu)
print(tv) # policzyłem wartość rezydualną (terminal value) na koniec 2030 roku
```

```python
dyskontowane_tv = tv / (1 + wacc) ** 5
print(dyskontowane_tv) # wartość rezydualna jest wyrażona "na koniec 2030", więc dyskontujemy ją tą samą potęgą co ostatni rok FCF (rok 5), żeby sprowadzić ją do dzisiejszej wartości
```

```python
enterprise_value = sum(zdyskontowane_fcf_5lat) + dyskontowane_tv
print(enterprise_value) # obliczyłem enterprise value jako sumę zdyskontowanych FCF z 5 lat i zdyskontowanej wartości rezydualnej
```

```python
gotowka = 954814
dlug_netto = dlug - gotowka
print(dlug_netto)#obliczyłem dług netto jako różnicę długu i gotówki, co ciekawe dług wychodzi ujemny co oznacza, że firma ma więcej gotówki niż długu, co jest pozytywnym sygnałem dla inwestorów.
```

```python
equity_value = enterprise_value - dlug_netto
print(equity_value) # obliczyłem equity value jako różnicę enterprise value i długu netto, co daje wartość kapitału własnego firmy
```

```python
ilosc_akcji = 980400000 # realna ilosc akcji dino
cena_akcji = equity_value / ilosc_akcji * 1000
print(cena_akcji) # obliczyłem cenę akcji jako wartość kapitału własnego podzieloną przez ilość akcji, pomnożoną przez 1000, aby uzyskać cenę jednej akcji w złotówkach.
```
